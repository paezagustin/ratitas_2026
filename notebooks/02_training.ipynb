```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
from ultralytics import YOLO

import dagshub, mlflow, torch
```

```python
os.environ['YOLO_MLFLOW'] = 'true'
dagshub.init(repo_owner = 'paezagustin', repo_name = 'ratitas_2026', mlflow=True)
#mlflow.set_tracking_uri('https://dagshub.com/paezagustin/ratitas_2026.mlflow')
#mlflow.set_experiment('Ratas_Segmentation_640x640')

model = YOLO("C:/Users/paez/Documents/Doct_2026/ratitas_2026/models/yolo26m-seg.pt")
```

```python
dataset_path = "C:/Users/paez/Documents/Doct_2026/ratitas_2026/data/processed/yolo_dataset_640x640/dataset.yaml"


results = model.train(
        data=dataset_path,
        epochs=1,          
        imgsz= 640,
        batch=16,
        workers=0,
        device=0,
        conf = 0.5,
        
        # --- DATA AUGMENTATION  ---
        degrees=180.0,       
        translate=0.2,      
        scale=0.5,          
        shear=0.0,          

        # --- Control de Entrenamiento ---
        patience=15,         
        save=True,          
        project="Ratas_Segmentation_640x640",
        name="seg_histologia_v2",
        exist_ok=True,
        visualize = True,
    )
```
